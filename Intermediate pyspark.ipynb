### UNION

### Preparing DataFrames

```python
data1 = [('1','kad'),
        ('2','sid')]
schema1 = 'id STRING, name STRING' 

df1 = spark.createDataFrame(data1,schema1)

data2 = [('3','rahul'),
        ('4','jas')]
schema2 = 'id STRING, name STRING' 

df2 = spark.createDataFrame(data2,schema2)
```

```python
df1.display()
df2.display()
```

```python
df1.union(df2).display()
```

```python
data1 = [('kad','1',),
        ('sid','2',)]
schema1 = 'name STRING, id STRING' 

df1 = spark.createDataFrame(data1,schema1)

df1.display()
```

```python
df1.union(df2).display()
```

### Union By namex

```python
df1.unionByName(df2).display() 
```

### String Funtions
#### INITCAP -- initcap() converts the first letter of each word in a string to uppercase and the remaining letters to lowercase.
#### UPPER()
#### lOWER() 

```python
from pyspark.sql.functions import *
df.select(initcap('Item_Type')).display()
```

```python
df.select(lower('Item_Type')).display()
```

```python
df.select(upper('Item_Type').alias('Upper_Items')).display()
```

### Date Functions
##### 1. CURRENT_DATE()
##### 2. DATE_ADD()
##### 3. DATE_SUB()

```python
df = df.withColumn('curr_date',current_date())

df.display()
```

```python
df = df.withColumn('week_after',date_add('curr_date',7))

df.display()
```

```python
df.withColumn('week_before',date_sub('curr_date',7)).display()
```

```python
df = df.withColumn('week_before',date_add('curr_date',-7)) 

df.display()
```

### Date Diff

```python
df = df.withColumn('datediff',datediff('week_after','curr_date'))

df.display()
```

### Date_format()

```python
df = df.withColumn('week_before',date_format('week_before','dd-MM-yyyy'))

df.display()
```

### Handling Nulls

##### Dropping Nulls

###### 1. Dropping records which are having null in all the columns

```python
df.dropna('all').display()
```

##### 2. Dropping records which have null in any one or more columns

```python
df.dropna('any').display()
```

### 3. Drop all those records which are having null in this particular column

```python
df.dropna(subset=['Outlet_Size']).display()
```

### Filling NUll Values

##### 1. fill null values in all the records accross the dataframe

```python
df.fillna('NotAvailable').display()
```

##### 2. Filling null values of this particular column

```python
df.fillna('NotAvailable',subset=['Outlet_Size']).display()
```

### SPLIT and INDEXING

##### SPLIT - split the value/record of the particular column to a list

```python
df.withColumn('Outlet_Type',split('Outlet_Type',' ')).display()
```

##### Indexing

```python
df.withColumn('Outlet_Type',split('Outlet_Type',' ')[1]).display()
```

### Explode
###### if you have a record and in one column it has value as list then it will create seperte more records for each value in the list

```python
df_exp = df.withColumn('Outlet_Type',split('Outlet_Type',' '))

df_exp.display()
```

```python
df_exp.withColumn('Outlet_Type',explode('Outlet_Type')).display()
```

```python
df_exp.display()
```

### Array contains

### COLLECT_LIST -- works like concat

```python
data = [('user1','book1'),
        ('user1','book2'),
        ('user2','book2'),
        ('user2','book4'),
        ('user3','book1')]

schema = 'user string, book string'

df_book = spark.createDataFrame(data,schema)

df_book.display()
```

```python
from pyspark.sql.functions import *
df_book.groupBy('user').agg(collect_list('book')).display()
```
