```python
df = spark.read.format('csv').options(header='true', inferSchema='true').load("/Volumes/workspace/default/my_volume/BigMart Sales.csv")
df.display()
```

### Data Writing

##### CSV

```python
df.write.format('csv')\
        .save('/Volumes/workspace/default/my_volume/data.csv')
```

##### Append

```python
df.write.format('csv')\
        .mode('append')\
        .save('/Volumes/workspace/default/my_volume/data.csv')
```

```python
df.write.format('csv')\
        .mode('append')\
        .option('path','/Volumes/workspace/default/my_volume/data.csv')\
        .save()
```

##### Over Write

```python
df.write.format('csv')\
.mode('overwrite')\
.option('path','/Volumes/workspace/default/my_volume/data.csv')\
.save()
```

##### Error

```python
df.write.format('csv')\
.mode('error')\
.option('path','/Volumes/workspace/default/my_volume/data.csv')\
.save()
```

##### Ignore

```python
df.write.format('csv')\
.mode('ignore')\
.option('path','/Volumes/workspace/default/my_volume/data.csv')\
.save()
```

### parquet

```python
df.write.format('parquet')\
.mode('overwrite')\
.option('path','/Volumes/workspace/default/my_volume/data.csv')\
.save()
```

### Table

```python
df.write.format('delta')\
.mode('overwrite')\
.saveAsTable('my_table')
```
