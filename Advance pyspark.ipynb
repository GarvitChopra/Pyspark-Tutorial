```python
df = spark.read.format('csv').options(header='true', inferSchema='true').load("/Volumes/workspace/default/my_volume/BigMart Sales.csv")
df.display()
```

### PIVOT

```python
from pyspark.sql.functions import *
df.groupBy('Item_Type').pivot('Outlet_Size').agg(avg('Item_MRP')).display()
```

### WHEN-OTHERWISE

```python
from pyspark.sql.functions import *
df = df.withColumn('veg_flag',when(col('Item_Type')=='Meat','Non-Veg').otherwise('Veg'))
```

```python
df.display()
```

```python
df.withColumn('veg_exp_flag',when(((col('veg_flag')=='Veg') & (col('Item_MRP')<100)),'Veg_Inexpensive').when((col('veg_flag')=='Veg') & (col('Item_MRP')>100),'Veg_Expensive').otherwise('Non_Veg')).display() 
```

JOINS

```python
dataj1 = [('1','gaur','d01'),
          ('2','kit','d02'),
          ('3','sam','d03'),
          ('4','tim','d03'),
          ('5','aman','d05'),
          ('6','nad','d06')] 

schemaj1 = 'emp_id STRING, emp_name STRING, dept_id STRING' 

df1 = spark.createDataFrame(dataj1,schemaj1)

dataj2 = [('d01','HR'),
          ('d02','Marketing'),
          ('d03','Accounts'),
          ('d04','IT'),
          ('d05','Finance')]

schemaj2 = 'dept_id STRING, department STRING'

df2 = spark.createDataFrame(dataj2,schemaj2)
```

```python
df1.display()
df2.display()
```

### INNER JOIN

```python
df1.join(df2, df1['dept_id']==df2['dept_id'],'inner').display()
```

### Left Join

```python
df1.join(df2,df1['dept_id']==df2['dept_id'],'left').display()
```

### Right Join

```python
df1.join(df2,df1['dept_id']==df2['dept_id'],'right').display()
```

### Anti Join
##### records which are not getting matched with other dataframe

```python
df1.join(df2,df1['dept_id']==df2['dept_id'],'anti').display()
```
