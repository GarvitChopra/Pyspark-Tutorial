### SELECT

```python
df = spark.read.format('csv').options(header='true', inferSchema='true').load("/Volumes/workspace/default/my_volume/BigMart Sales.csv")
df.display()
```

```python
from pyspark.sql.functions import *
df.select(col('Item_Identifier'),col('Item_Weight'),col('Item_Fat_Content')).display()
```

### ALIAS

```python
df.select(
    col("Item_Identifier").alias("Item ID")
).display()
```

### FILTER AND WHERE

```python
df.filter(col('Item_Fat_Content')== 'Regular').display()
```

```python
df.filter((col('Item_Type') == 'Soft Drinks') & (col('Item_Weight') < 10)).display()
```

```python
df.filter((col("Outlet_Size").isNull()) & col('Outlet_Location_Type').isin(['Tier 1','Tier 2'])).display()
```

### with Column Renamed ---- change the column name at dataframe level

```python
df.withColumnRenamed('Item_Weight','Item Wt').display()
```

### withColumn --- make a new column or modify a existing column

## create new column

```python
from pyspark.sql.functions import lit
df = df.withColumn('flag',lit('new'))
```

```python
df.display()
```

```python
df.withColumn('multiply',col('Item_Outlet_Sales')*col('Item_MRP')).display()
```

### modify existing column

```python
from pyspark.sql.functions import regexp_replace
df.withColumn('Item_Fat_Content',regexp_replace(col('Item_Fat_Content'),'Low Fat','LF'))\
  .withColumn('Item_Fat_Content',regexp_replace(col('Item_Fat_Content'),'Regular','REG'))

df.display()
```

### Type Casting

```python
df.printSchema()
```

```python
from pyspark.sql.functions import *
df = df.withColumn('Item_Weight',col('Item_Weight').cast(StringType()))
```

```python
df.printSchema()
```

### Sort/Order By

```python
df.sort(col("Item_Weight").desc()).display()
```

```python
df.sort(col("Item_Visibility").asc()).display() 
```

### Sorting based on multiply column

```python
df.sort(['Item_Weight','Item_Visibility'],ascending = [0,0]).display()
```

```python
df.sort(['Item_Weight','Item_Visibility'],ascending = [0,1]).display()
```

### LIMIT

```python
df.limit(10).display()
```

### Drop -- used to drop columns

```python
df.drop('Item_Visibility').display()
```

```python
df.drop('Item_Visibility','Outlet_Type').display()
```

### Drop Duplicates  --- 1. drop duplicate rows 2. drop the duplicate items in the column

```python
df.dropDuplicates().display()
```

```python
df.dropDuplicates(subset=["Item_Type"]).display()
```

### df.distinct().display() returns all unique rows from your DataFrame, removing duplicate rows.

```python
df.distinct().display()
```
